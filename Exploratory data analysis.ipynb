```python
Q1 = df.TotalPrice.quantile(0.25)
Q3 = df.TotalPrice.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print('Q1:', Q1)
print('Q3:', Q3)
print('IQR:', IQR)
print('Lower Bound:', lower_bound)
print('Upper Bound:', upper_bound)
```

```python
outliers = df[(df['TotalPrice'] < lower_bound) | (df['TotalPrice'] > upper_bound)]
print('Number of outliers:', len(outliers))
```

```python
outliers[["OrderID","Date","Product","Quantity","UnitPrice","ItemsInCart","TotalPrice"]]
```

The detected TotalPrice outliers are concentrated among orders with the maximum quantity of 5 and high unit peices, suggesting that the outliers are driven by legitimate high-value purchases rather than obvious data-entry errors

```python
df['Year'] = df['Date'].dt.year
df['Year'].value_counts().sort_index()
```

2023 had the highest number of orders (510), followed by 2024 (459). 2025 has considerabling fewer orders (231)

```python
df.groupby('Year')['TotalPrice'].agg(["count", "sum", "mean", "median"])
```

```python
df.groupby("Year")['TotalPrice'].sum().plot(kind='bar', title='Total Revenue By Year', xlabel='Year', ylabel='Total Revenue')
```

```python
df['Month'] = df["Date"].dt.month

monthly_orders = df.groupby('Month')['OrderID'].count()

monthly_orders
```

```python
monthly_orders.plot(kind='line', marker='o',title='Monthly Order Trend',xlabel='Month',ylabel='Number of Orders')
```

June recorded the highest monthly order volume with 147 orders, while July and December recorded the lowest with 87 orders each

```python
monthly_revenue = df.groupby('Month')['TotalPrice'].sum()

monthly_revenue.sort_values(ascending= False)
```

```python
monthly_revenue.plot(kind='line', marker='o',title='Monthly Revenue Trend',xlabel='Month',ylabel='Revenue')
```

June has the highest monthly revenue (170,6161.13) September had the lowest revenue (69,321.65)

```python
correlation = df[["Quantity",'UnitPrice','ItemsInCart','TotalPrice']].corr().round(2)

correlation
```

```python
payment = df.groupby('PaymentMethod').agg(Orders=('OrderID','count'), Revenue=('TotalPrice','sum'), AverageOrderValue=('TotalPrice','mean')).sort_values('Orders', ascending=False)
payment['Orders'].plot(kind='bar')
plt.title('Orders By Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
```

Unit Price had the strongest positive correlation with Total Price (0.72) followed by Quantity (0.62)

```python
df.groupby('ReferralSource').agg(Orders=('OrderID','count'), Revenue=('TotalPrice','sum'), AverageOrderValue=('TotalPrice','mean')).sort_values('Orders', ascending=False)
```

```python
df.groupby('CouponCode').agg(Orders=('OrderID','count'), Revenue=('TotalPrice','sum'), AverageOrderValue=('TotalPrice','mean')).sort_values('Orders', ascending=False)
```

## Conclusion
The exploratory data analysis provided useful insights into the dataset's patterns,
distributions, trends, and outliers.

The dataset contains 1,200 orders. 2023 recorded the highest number of orders
and the highest annual revenue. June had the highest monthly order volume and
monthly revenue, while September recorded the lowest monthly order volume and
revenue.

The analysis also identified 8 outliers in TotalPrice using the IQR method.
These outliers were associated with high quantities and high unit prices and
did not appear to be obvious data-entry errors.

UnitPrice had the strongest positive relationship with TotalPrice, followed
by Quantity. Online was the most frequently used payment method, FREESHIP
was the most-used coupon, and Instagram was the leading referral source.

Overall, the analysis demonstrates how descriptive statistics, visualization,
trend analysis, correlation, and outlier detection can be used to transform
raw transactional data into meaningful insights.
